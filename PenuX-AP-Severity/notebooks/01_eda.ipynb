Set `DATA_PATH` and `TARGET_COLUMN` below before running.

```python
import os
import sys
sys.path.insert(0, '../src')

DATA_PATH = os.environ.get('DATA_PATH', '')  # Set this
TARGET_COLUMN = os.environ.get('TARGET_COLUMN', 'severe')

if not DATA_PATH:
    raise RuntimeError(
        'Set DATA_PATH to a sanitized dataset path before running.\n'
        'No dataset is bundled. Add a legally usable de-identified dataset to data/public_sanitized/.'
    )
```

```python
import pandas as pd
from penux_ap.datasets import load_dataset, detect_target_column
from penux_ap.preprocessing import summarize_missingness, infer_feature_types
from penux_ap.labels import describe_target, binarize_target

df = load_dataset(DATA_PATH)
print(f'Shape: {df.shape}')
df.head()
```

```python
target_col = TARGET_COLUMN or detect_target_column(df)
y = binarize_target(df[target_col])
print(describe_target(y))
```

```python
miss = summarize_missingness(df)
print(miss)
```

```python
feature_types = infer_feature_types(df, target_col)
print('Numeric:', feature_types['numeric'])
print('Categorical:', feature_types['categorical'])
```
